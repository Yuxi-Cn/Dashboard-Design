1. **Set index**: Set the 'DATETIME' column as the index.

2. **Calculate sum**: Calculate monthly, seasonal, annual sum for each source and carbon intensity.

3. **Decimal Precision**: Set decimal places to 2 for clear and concise data values.

4. **Unit Conversion**: Convert summed source generation from kWh to GW and summed carbon intensity from gCO2/kWh to kgCO2/kWh for enhancing readability.

5. **Calculate individual source percentages**: When summing data, compute source percentages by dividing each source's generation by the total generation, instead of summing up all percentages when operating '.sum()' function. This is to accurately represent their contributions.

```python
df = pd.read_csv('09-24_fuel_gb.csv')
```

```python
# convert the 'DATETIME' column to datetime
df['DATETIME'] = pd.to_datetime(df['DATETIME'])
# set the 'DATETIME' column as the index
df.set_index('DATETIME', inplace=True)

# resample the data to monthly intervals and sum the values
df_month = df.resample('M').sum()
# resample the data to seasonal intervals (3 months) and sum the values
df_season = df.resample('Q').sum()
# resample the data to annual intervals  and sum the values
df_annual = df.resample('Y').sum()
```

```python
print(df_annual.info())
print(df_annual.shape)
```

**Convert Data Units and Decimal Places & Differentiate Zero_carbon and Low_carbon**

```python
# define DataFrames
dfs = [df_month, df_season, df_annual]

# units conversion and precision adjustment
columns_to_convert = ['GAS', 'COAL', 'NUCLEAR', 'WIND', 'HYDRO', 'IMPORTS', 'BIOMASS', 'OTHER', 'SOLAR',
                      'STORAGE', 'GENERATION', 'LOW_CARBON', 'ZERO_CARBON', 'RENEWABLE', 'FOSSIL']

# replace the 'LOW_CARBON' data with the 'BIOMASS' data
# to ensure there is no overlap between each type of generation (such as "ZERO_CARBON" and LOW_CARBON")
for df in dfs:
    df['LOW_CARBON'] = df['BIOMASS']

# convert MWh to GWh
for df in dfs:
    df[columns_to_convert] = df[columns_to_convert] / 1000

# convert g/kWh to kg/kWh for carbon intensity
for df in dfs:
    df['CARBON_INTENSITY'] = df['CARBON_INTENSITY'] / 1000

# calculate percentages for source and generation categories
for df in dfs:
    source_columns = ['GAS', 'COAL', 'NUCLEAR', 'WIND', 'HYDRO', 'IMPORTS', 'BIOMASS', 'OTHER',
                      'SOLAR', 'STORAGE', 'GENERATION', 'LOW_CARBON', 'ZERO_CARBON', 'RENEWABLE', 'FOSSIL']
    for source in source_columns:
        perc_column = source + '_perc'
        df[perc_column] = (df[source] / df['GENERATION']) * 100

# round all columns to 2 decimal places
for df in dfs:
    df[:] = df.round(2)

# set GENERATION_perc to 100 for all years
for df in dfs:
    df['GENERATION_perc'] = 100
```

**The Unit of Data after the Conversion:**

*   Sources Generation Amount: **GWh**

*   Carbon Intensity: **kgCO2/kWh**

*   Source Percentage: **percent (%)**

**Convert the datetime index to the corresponding level (year, quarter, month)**

```python
# for df_annual (Year)
df_annual.index = df_annual.index.to_period('Y').to_timestamp()

# for df_season (Quarter)
df_season.index = df_season.index.to_period('Q').to_timestamp()

# for df_month (Month)
df_month.index = df_month.index.to_period('M').to_timestamp()
```

##### **2. Quarterly Variations**

```python
import plotly.graph_objects as go
import pandas as pd

# Assume df_season is your DataFrame equivalent to seasonal_sum
# Plot the generation data

fig = go.Figure()

# Add the main generation line plot
fig.add_trace(go.Scatter(
    x=df_season.index,
    y=df_season['GENERATION'],
    mode='lines+markers',
    marker=dict(size=8, symbol='circle', color='blue'),
    line=dict(color='blue', width=2),
    name='Generation',
    hovertemplate='Generation: %{y:.2f} GWh<br>Date: %{x|%Y-%m-%d}<extra></extra>'
))

# Annotate the first and last fourth quarter (2009-10-01 and 2023-10-01)
first_fourth_quarter_index = df_season.index.get_loc(pd.Timestamp('2009-10-01'))
last_fourth_quarter_index = df_season.index.get_loc(pd.Timestamp('2023-10-01'))

fig.add_annotation(
    x=df_season.index[first_fourth_quarter_index],
    y=df_season['GENERATION'].iloc[first_fourth_quarter_index],
    text=f"{df_season['GENERATION'].iloc[first_fourth_quarter_index]:.2f}",
    showarrow=True,
    arrowhead=1,
    ax=0,
    ay=-40,
    font=dict(color='black', size=12)
)

fig.add_annotation(
    x=df_season.index[last_fourth_quarter_index],
    y=df_season['GENERATION'].iloc[last_fourth_quarter_index],
    text=f"{df_season['GENERATION'].iloc[last_fourth_quarter_index]:.2f}",
    showarrow=True,
    arrowhead=1,
    ax=0,
    ay=-40,
    font=dict(color='black', size=12)
)

# Highlight every second and third quarter (specifically, 03-31 to 09-30), including 2023
for year in range(2009, 2024):  # Adjusting for 2023 data
    highlight_color = 'lightblue' if year == 2023 else 'gray'
    opacity_value = 0.7 if year == 2023 else 0.4

    fig.add_vrect(
        x0=pd.Timestamp(f'{year}-03-31'),
        x1=pd.Timestamp(f'{year}-06-30'),
        fillcolor=highlight_color,
        opacity=opacity_value,
        layer='below',
        line_width=0,
    )
    fig.add_vrect(
        x0=pd.Timestamp(f'{year}-06-30'),
        x1=pd.Timestamp(f'{year}-09-30'),
        fillcolor=highlight_color,
        opacity=opacity_value,
        layer='below',
        line_width=0,
    )

# Update layout for better aesthetics and proper x-axis ticks
fig.update_layout(
    title='Seasonal Trends in Total Electricity Generation',
    xaxis=dict(
        title='Year',
        titlefont=dict(size=14),
        tickfont=dict(size=12),
        showline=True,  # Add bottom border
        linewidth=2,
        linecolor='black',
        dtick="M24",  # Set the x-axis to show every 2 years
        tickformat="%Y"  # Format the tick labels to display years
    ),
    yaxis=dict(
        title='Generation Amount (GWh)',
        titlefont=dict(size=14),
        tickfont=dict(size=12),
        showline=True,  # Add left border
        linewidth=2,
        linecolor='black'
    ),
    hovermode="x unified",  # Show all hover info in a unified box
    plot_bgcolor='white',
    margin=dict(t=50, l=50, r=50, b=50),
    showlegend=False
)

# Update x-axis to rotate labels for better readability
fig.update_xaxes(tickangle=45)

# Add a comment comparing the first and last fourth quarter generation difference
first_fourth_quarter = df_season['GENERATION'].loc[pd.Timestamp('2009-10-01')]
last_fourth_quarter = df_season['GENERATION'].loc[pd.Timestamp('2023-10-01')]
generation_diff = last_fourth_quarter - first_fourth_quarter

# Position the generation difference annotation above the last data point annotation
fig.add_annotation(
    x=pd.Timestamp('2023-10-01'),
    y=last_fourth_quarter + 5,  # Slightly above the last data point annotation
    text=f"Generation Difference (2009-2023 Q4): {generation_diff:.2f} GWh",
    showarrow=False,
    font=dict(color='red', size=12),
    bgcolor="white",
    opacity=0.8,
    xanchor='center',
    yanchor='top'
)

# Display the plot
fig.show()
```

```python
import plotly.graph_objs as go
import plotly.subplots as sp
import pandas as pd

# Example data: Extract data for 2009 and 2023
first_year = df_annual.iloc[0]
last_year = df_annual.iloc[-1]

# Define the energy categories
fossil_sources = ["GAS", "COAL", "NUCLEAR", "OTHER"]
renewable_sources = ["WIND", "HYDRO", "BIOMASS", "SOLAR"]

# Prepare data for the first year with "2009" at the center
labels_first = ["2009", "Fossil", "Renewable"] + fossil_sources + renewable_sources
parents_first = ["", "2009", "2009"] + ["Fossil"]*len(fossil_sources) + ["Renewable"]*len(renewable_sources)
values_first = [
    first_year[fossil_sources].sum() + first_year[renewable_sources].sum(),  # Total Energy
    first_year[fossil_sources].sum(),  # Fossil Total
    first_year[renewable_sources].sum(),  # Renewable Total
] + list(first_year[fossil_sources]) + list(first_year[renewable_sources])

# Prepare data for the last year with "2023" at the center
labels_last = ["2023", "Fossil", "Renewable"] + fossil_sources + renewable_sources
parents_last = ["", "2023", "2023"] + ["Fossil"]*len(fossil_sources) + ["Renewable"]*len(renewable_sources)
values_last = [
    last_year[fossil_sources].sum() + last_year[renewable_sources].sum(),  # Total Energy
    last_year[fossil_sources].sum(),  # Fossil Total
    last_year[renewable_sources].sum(),  # Renewable Total
] + list(last_year[fossil_sources]) + list(last_year[renewable_sources])

# Create the subplot layout
fig = sp.make_subplots(rows=1, cols=2, specs=[[{'type': 'sunburst'}, {'type': 'sunburst'}]])

# Add the first year sunburst to the first subplot
fig.add_trace(go.Sunburst(
    labels=labels_first,
    parents=parents_first,
    values=values_first,
    branchvalues="total",
    hoverinfo="label+percent entry",
), row=1, col=1)

# Add the last year sunburst to the second subplot
fig.add_trace(go.Sunburst(
    labels=labels_last,
    parents=parents_last,
    values=values_last,
    branchvalues="total",
    hoverinfo="label+percent entry",
), row=1, col=2)

# Update layout
fig.update_layout(
    title_text="Energy Distribution: 2009 vs 2023",
    paper_bgcolor='rgba(255, 255, 255, 1)',  # White background for better contrast
    margin=dict(t=50, l=25, r=25, b=25),
    sunburstcolorway=["#636EFA", "#EF553B", "#00CC96", "#AB63FA", "#FFA15A", "#19D3F3", "#FF6692", "#B6E880", "#FF97FF", "#FECB52"],
    uniformtext=dict(minsize=10, mode='hide')
)

fig.show()
```
